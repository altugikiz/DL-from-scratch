```python
import numpy as np
import math
```

```python
def basic_sigmoid(x):
    """
    x: a real number
    return: sigmoid(x)
    """
    s = 1 / (1 + math.exp(-x))
    return s
```

```python
print(basic_sigmoid(1))
```

```python
def sigmoid(x):
    """
    x: scalar or numpy array
    return: sigmoid(x), same shape as x
    """
    s = 1 / (1 + np.exp(-x))
    return s
```

```python
X = np.array([1, 2, 3])
print(sigmoid(X))
```

```python
def sigmoid_derivative(x):
    """
    x: scalar or numpy array
    return: gradient of sigmoid at x
    """
    s = sigmoid(x)
    ds = s * (1 - s)
    return ds
```

```python
x = np.array([1, 2, 3])
print(sigmoid_derivative(x))
```

```python
def image2vector(image):
    """
    image: numpy array of shape (length, height, depth)
    return: vector of shape (length*height*depth, 1)
    """
    v = image.reshape(-1, 1)
    return v
```

```python
image = np.array([[[0.67, 0.03], [0.34, 0.61], [0.22, 0.35]],
                   [[0.72, 0.14], [0.83, 0.99], [0.60, 0.42]],
                   [[0.19, 0.51], [0.28, 0.13], [0.44, 0.90]]])
print(image2vector(image).shape)
```

```python
def normalize_rows(x):
    """
    x: numpy matrix of shape (n, m)
    return: normalized version of x, same shape
    """
    x_norm = np.linalg.norm(x, axis=1, keepdims=True)
    x = x / x_norm
    return x
```

```python
x = np.array([[0, 3, 4], [1, 6, 4]])
print(normalize_rows(x))
```

```python
def softmax(x):
    """
    x: numpy matrix of shape (n, m)
    return: softmax(x), same shape as x
    """
    x_exp = np.exp(x)
    x_sum = np.sum(x_exp, axis=1, keepdims=True)
    s = x_exp / x_sum 
    return s
```

```python
x = np.array([[9, 2, 5, 0, 0], [7, 5, 0, 0, 0]])
print(softmax(x))
```

```python
def L1(yhat, y):
    """
    yhat: predicted labels, shape (m,)
    y: true labels, shape (m,)
    return: L1 loss
    """
    loss = np.sum(np.abs(y - yhat))
    return loss
```

```python
yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])
print(L1(yhat, y))
```

```python
def L2(yhat, y):
    """
    yhat: predicted labels, shape (m,)
    y: true labels, shape (m,)
    return: L2 loss
    """
    diff = y - yhat
    loss = np.dot(diff, diff)
    return loss
```

```python
yhat = np.array([.9, 0.2, 0.1, .4, .9])
y = np.array([1, 0, 0, 1, 1])
print(L2(yhat, y))
```
